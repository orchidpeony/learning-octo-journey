# 1. Buatlah sebuah fungsi untuk mendapatkan nilai terbesar dari input 3 (tiga) buah bilangan. Catatan: tidak boleh memanfaatkan fungsi max() atau built-in function lainnya.

```python
def nilai_terbesar(a,b,c):
  terbesar = a
  if b > terbesar:
    terbesar = b
  if c > terbesar:
    terbesar = c
  return terbesar

print(nilai_terbesar(20,25,30))
```

#2. Buatlah sebuah fungsi untuk mendapatkan nilai rata-rata dari input sejumlah n bilangan dalam sebuah list. Catatan: tidak boleh menggunakan fungsi sum(), mean() atau built-in function lainnya.

```python
def rata_rata(MyData):
  total = 0
  jumlah = 0
  for angka in MyData:
   total = angka + total
   jumlah = jumlah + 1
  hasil = total / jumlah
  return hasil
print(rata_rata([20,30,40,50]))
```

#3. Buatlah sebuah fungsi yang menghitung banyaknya bilangan genap dan bilangan ganjil dari sejumlah elemen di dalam sebuah list. Output berupa sebuah tuple dengan dua elemen: elemen pertama adalah banyanya bilangan genap, elemen kedua adalah banyaknya bilangan ganjil.

```python
def jumlah_bilangan(MyData):
  genap = 0
  ganjil = 0
  for angka in MyData:
    if angka % 2 == 0:
      genap = genap + 1
    else:
      ganjil = ganjil + 1
  return (genap,ganjil)
print(jumlah_bilangan([19,23,24,25,26,30,37]))
```

#4. Diberikan sebuah string yang merupakan campuran dari huruf kecil dan huruf kapital. Buatlah sebuah fungsi yang mengubah susunan huruf dalam string tersebut supaya huruf kecil muncul terlebih dahulu baru diikuti huruf kapital

```python
def susunan_string(BurungKaliBri):
  huruf_besar = ""
  huruf_kecil = ""
  for huruf in BurungKaliBri:
    if huruf .isupper():
       huruf_besar = huruf_besar + huruf
    elif huruf .islower():
      huruf_kecil = huruf_kecil + huruf
  return huruf_kecil + huruf_besar

print(susunan_string("BurungKaliBri"))
```

#5. Diberikan suatu kata dalam string, buatlah fungsi yang menghitung banyaknya kemunculan setiap karakter (huruf/angka/symbol) dalam kata tersebut. Hasil kembaliannya harus berupa dictionary. Dilarang menggunakan fungsi built-in count() dari Python

```python
def jumlah_karakter(burungkalibri):
  karakter = {}
  for huruf in burungkalibri:
    if  huruf in karakter:
      karakter[huruf] = +1
    else:
      karakter[huruf] = 1
  return karakter
print(jumlah_karakter("burungkalibri"))
```

#6. Buatlah sebuah fungsi dengan nama show_employee() berdasarkan ketentuan berikut:

· Fungsi harus menerima nama dan gaji employee dan menampilkan keduanya di layar.

· Jika input gajinya kosong saat pemanggilan fungsi, isikan nilai default 2000000 untuk gaji.

· Fungsi juga mengembalikan suatu dictionary dengan key-nya adalah “nama” dan “gaji”, value-nya adalah nilai nama dan gajinya

```python
def show_employee(nama, gaji = 2000000):
  print("nama:" , nama)
  print("gaji:", gaji)
  return {"nama": nama, "gaji": gaji}
print(show_employee("medina"))
print(show_employee("wulan", 3500000))
print(show_employee("dani"))
print(show_employee("maria", 4000000))
```
